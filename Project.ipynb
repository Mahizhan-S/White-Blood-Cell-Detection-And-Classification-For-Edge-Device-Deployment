```python
import os

label_dir = "WBCDD2/labels/train"
class_ids = set()

for f in os.listdir(label_dir):
    with open(os.path.join(label_dir, f)) as file:
        for line in file:
            class_ids.add(int(line.split()[0]))

print("Class IDs found in labels:", sorted(class_ids))
```
